```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
numerical_cols = [
    'Distance_km',
    'Shipment_Volume_m3',
    'Package_Weight_kg',
    'Number_of_Stops',
    'Driver_Experience_years',
    'Delivery_Time_min',
    'Transportation_Cost_INR'
]
```

```python
print("MEAN (Average)")
display(df[numerical_cols].mean().round(2))
```

```python
display(df[numerical_cols].median().round(2))
```

```python
display(df[numerical_cols].std().round(2))
```

```python
print("MAXIMUM VALUES")
display(df[numerical_cols].max())
```

```python
print("Delivery delay counts:")
print(df['Delivery_Delayed'].value_counts())
```

```python

print("\nDelivery delay percentage:")
print(
    (df['Delivery_Delayed'].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
print("Average transportation cost:",round(df['Transportation_Cost_INR'].mean(), 2), "INR")
```

```python
(df['Delivery_Delayed'].value_counts(normalize=True) * 100).round(2)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Delivery_Time_min', bins=25, kde=True)
plt.title('Distribution of Delivery Time')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Number of Deliveries')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Transportation_Cost_INR', bins=25, kde=True)
plt.title('Distribution of Transportation Cost')
plt.xlabel('Transportation Cost (INR)')
plt.ylabel('Number of Deliveries')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Distance_km',
    y='Delivery_Time_min',
    hue='Traffic_Level',
    alpha=0.7
)
```

```python
plt.title('Distance vs Delivery Time by Traffic Level')
plt.xlabel('Distance (km)')
plt.ylabel('Delivery Time (minutes)')
plt.tight_layout()
plt.show()
```

```python

# 1. Correlation heatmap
numerical_df = df.select_dtypes(include='number')
correlation = numerical_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)


```

```python
plt.title('Correlation Heatmap of Logistics Variables')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Traffic_Level', y='Delivery_Time_min')
plt.title('Delivery Time by Traffic Level')
plt.xlabel('Traffic Level')
plt.ylabel('Delivery Time (minutes)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Vehicle_Type', y='Transportation_Cost_INR')
plt.title('Transportation Cost by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Transportation Cost (INR)')
plt.tight_layout()
plt.show()
```

```python
print("Traffic-wise performance:")
display(
    df.groupby('Traffic_Level')[
        ['Delivery_Time_min', 'Transportation_Cost_INR']
    ].mean().round(2)
)
```

```python
print("Weather-wise performance:")
display(
    df.groupby('Weather')[
        ['Delivery_Time_min', 'Transportation_Cost_INR']
    ].mean().round(2)
)
```

```python
print("Vehicle-wise performance:")
display(
    df.groupby('Vehicle_Type')[
        ['Delivery_Time_min', 'Transportation_Cost_INR']
    ].mean().round(2)
)
```

```python
print("Delay rate by traffic (%):")
display(
    (df.groupby('Traffic_Level')['Delivery_Delayed'].mean() * 100)
    .round(2)
)
```

```python



# Delivery delay count by traffic
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x='Traffic_Level',
    hue='Delivery_Delayed'
)
plt.title('Delayed vs Non-delayed Deliveries by Traffic')
plt.xlabel('Traffic Level')
plt.ylabel('Number of Deliveries')
plt.legend(title='Delivery Delayed (0=No, 1=Yes)')
plt.tight_layout()
plt.show()

# Average transportation cost by vehicle
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x='Vehicle_Type',
    y='Transportation_Cost_INR',
    errorbar=None
)
plt.title('Average Transportation Cost by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Average Cost (INR)')
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='Weather',
    y='Delivery_Time_min',
    errorbar=None
)

plt.title('Average Delivery Time by Weather')
plt.xlabel('Weather Condition')
plt.ylabel('Average Delivery Time (minutes)')
plt.tight_layout()
plt.show()
```

```python

print("TOTAL DELIVERIES:", len(df))

print("\nAVERAGE DELIVERY TIME:")
print(round(df['Delivery_Time_min'].mean(), 2), "minutes")

print("\nAVERAGE TRANSPORTATION COST:")
print(round(df['Transportation_Cost_INR'].mean(), 2), "INR")

print("\nDELAY RATE:")
print(round(df['Delivery_Delayed'].mean() * 100, 2), "%")

print("\nTRAFFIC-WISE AVERAGE DELIVERY TIME:")
print(
    df.groupby('Traffic_Level')['Delivery_Time_min']
    .mean().round(2)
)

print("\nWEATHER-WISE AVERAGE DELIVERY TIME:")
print(
    df.groupby('Weather')['Delivery_Time_min']
    .mean().round(2)
)

print("\nVEHICLE-WISE AVERAGE COST:")
print(
    df.groupby('Vehicle_Type')['Transportation_Cost_INR']
    .mean().round(2)
)
```
